# Folium 지도에 좌표 표시하기
- 위도와 경도로 지도를 만들고 마커와 팝업을 추가한 뒤 HTML로 저장하는 실습입니다.
- 공식 문서: https://python-visualization.github.io/folium/latest/getting_started.html

```python
# Folium이 설치되지 않은 경우 터미널에서 아래 명령으로 설치합니다.
# 이미 정상 작동한다면 실행할 필요가 없습니다.
# uv add folium
```

```python
# Folium은 위도와 경도를 웹 기반의 대화형 지도에 표시하는 라이브러리입니다.
import folium
```

```python
# 서울시청 좌표를 최초 중심 위치로 하는 지도 객체를 만듭니다.
# location은 [위도, 경도] 순서이며 zoom_start가 클수록 더 확대됩니다.
folium_map = folium.Map(
    location=[37.5662952, 126.9779451],
    zoom_start=16
)
```

```python
# Marker로 서울시청 위치에 핀을 만들고 add_to()로 지도에 추가합니다.
folium.Marker(
    location=[37.5662952, 126.9779451]
).add_to(folium_map)
```

```python
# Popup은 마커를 클릭했을 때 표시할 정보창이며 max_width는 최대 너비입니다.
popup = folium.Popup('서울시청 위치', max_width=300)

# 팝업을 연결한 마커를 같은 서울시청 좌표에 추가합니다.
# 앞 셀의 마커와 위치가 같아서 지도에서는 두 마커가 겹쳐 보일 수 있습니다.
folium.Marker(
    location=[37.5662952, 126.9779451],
    popup=popup
).add_to(folium_map)
```

```python
# 셀의 마지막 줄에 지도 객체를 적으면 Jupyter에 대화형 지도가 출력됩니다.
folium_map
```

```python
# 폴더 생성 기능을 사용하기 위해 Python 표준 라이브러리 os를 불러옵니다.
import os

# map_data 폴더가 없으면 만들고, 이미 있으면 그대로 사용합니다.
os.makedirs('map_data', exist_ok=True)

# 완성된 지도를 브라우저에서 열 수 있는 HTML 파일로 저장합니다.
folium_map.save('map_data/folium_map.html')
```
